# Terraform para Artifact Registry y Google Cloud Functions.



Esta sesion formaliza el aprovisionamiento de destino productivo en GCP para que el pipeline no dependa de infraestructura creada manualmente.



## Objetivos.



- Modelar recursos base para una ruta de despliegue en GCP.

- Relacionar artefactos versionados con el entorno de ejecucion en prod.

- Entender el papel de Google Artifact Registry y Cloud Functions como runtime gestionado.


## Puente rapido de Terraform (si partes de cero).

Antes de modelar recursos en GCP, valida este mapa mental:

- Terraform describe infraestructura deseada, no pasos manuales.
- `plan` te deja revisar cambios antes de ejecutar.
- `apply` solo debe correr con aprobacion y contexto de entorno.
- El `state` remoto evita colisiones cuando trabajan varias personas o pipelines.

En esta sesion nos enfocamos en traducir componentes de despliegue (registry, service account y function) a recursos declarativos versionados.

## Recursos tipicos.



Una ruta tipica modela al menos:



- un repositorio de artefactos e imagenes en Google Artifact Registry;

- una cuenta de servicio con permisos acotados;

- una funcion en Google Cloud Functions con configuracion por entorno;

- variables y secretos separados para test y prod.



El objetivo es que la entrega continua tenga un destino reproducible y no dependiente de configuracion manual.



## Relacion entre artefacto, runtime gestionado y entorno.



En esta ruta, prod vive en GCP con control de identidad federada. El entorno Linux de test se trata en cuadernos posteriores como ruta didactica de operacion sobre host.



Artifact Registry permite mantener trazabilidad entre commit, build y despliegue sin reconstruir imagen por entorno.

## Puente con runners y despliegue.



La decision de runners se mantiene consistente con el curso:



- hosted como opcion por defecto para orquestar pipeline;

- self-hosted solo cuando una etapa requiere red privada o controles adicionales.



Esto permite sostener un mismo proceso para test Linux y prod GCP sin duplicar logica de build.
