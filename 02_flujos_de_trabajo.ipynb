# Flujos de trabajo.

Como se vio en capítulos previos, el repositorio de código y sus políticas de acceso forman parte de la superficie de ataque de un *pipeline DevSecOps*. La elección del flujo de trabajo no es solo una decisión de productividad del equipo: determina quién puede introducir cambios en qué ramas, cuántos puntos de revisión existen antes de que un cambio llegue a producción, y qué tan fácil resulta para un atacante introducir código malicioso sin pasar por un control técnico verificable.

A partir de la experiencia de varios grupos de trabajo se han adoptado varios esquemas de flujo de trabajo. Los siguientes son los más populares.

Atlassian ha publicado una comparación de flujos de trabajo la cual puede ser consultada en:

https://www.atlassian.com/git/tutorials/comparing-workflows

## Objetivos de la sesión

Al finalizar esta sesión podrás:

- comparar los principales modelos de flujo de trabajo con *Git* y explicar sus *trade-offs* de seguridad;
- identificar qué controles de revisión y *quality gates* habilita o bloquea cada flujo;
- seleccionar un flujo de trabajo adecuado para equipos que usan *pull requests* y *CI/CD* en *GitHub*.

## Flujo centralizado.

Este flujo de trabajo es común en repositorios cliente-servidor, en los que solo existe la rama única.

Desde la perspectiva *DevSecOps*, el flujo centralizado presenta el riesgo más alto: cualquier desarrollador con acceso puede empujar código directamente a la rama principal sin revisión previa. No existe un punto de control automatizado antes de que el cambio sea visible y ejecutable. Este modelo hace imposible implementar *quality gates* como condición de integración.

## Flujo por "*feature*".

Este flujo cuenta con una rama principal y una o más ramas de funcionalidades. Cada funcionalidad es independiente de las otras.

https://www.atlassian.com/git/tutorials/comparing-workflows/feature-branch-workflow

El flujo por *feature* (*rama de funcionalidad*) es la base del modelo de revisión en *GitHub*: cada cambio vive en una rama corta y se integra a través de un *pull request*. Ese *pull request* es el punto donde los *quality gates* automáticos de *CI/CD* —*lint*, pruebas, análisis de seguridad— se ejecutan como condición de *merge*. Si una rama contiene código problemático introducido desde una dependencia comprometida o un *fork* malicioso, ese código no llega a la rama principal sin pasar por revisión y *checks* automatizados.

<img src="img/02/feature.png" width="70%">

## *Gitflow*.

[*Gitflow*](https://nvie.com/posts/a-successful-git-branching-model/) es un modelo de flujo de trabajo propuesto por Vincent Driessen.

Este flujo de trabajo considera el desarrollo de las ramas:

* **master/main**, la cual es la rama principal y a la que solo puede agregarse código totalmente probado. Esta rama contiene el código productivo.
* **develop**, la cual corresponde a la rama en la que se realiza el desarrollo cotidiano.
* **hotfixes**, la cual se utiliza exclusivamente para atender correcciones de emergencia (*bugs*) y que serán añadidas a la rama **master/main**.
* **release branch**, la cual permite preparar liberaciones calendarizadas de código proveniente de **develop**. En esta rama se realizan los análisis y pruebas que garanticen que el código puede ser incluido en **master/main**.
* **feature branches** las cuales corresponden a funcionalidades que se irán integrando con el tiempo a **develop**.

https://www.atlassian.com/git/tutorials/comparing-workflows/gitflow-workflow

*Gitflow* protege la rama principal con varias capas de integración intermedias, lo que reduce el riesgo de introducir código no validado en producción. Sin embargo, su complejidad puede dificultar la aplicación uniforme de *policies* de seguridad sobre todas las ramas activas, especialmente cuando existen muchas *feature branches* de larga duración que no integran los *checks* actualizados del *pipeline*.

## *Trunk-based development*.

Cuando una organización ha implementado y madurado su proceso de *DevOps* puede utilizar un flujo de trabajo similar al flujo centralizado en el que cada desarrollador del equipo realiza un *commit* de su código al final del día y dicho *commit* desencadena una serie de acciones automatizadas que lo validan y lo preparan para ser entregado o desplegado en producción.

* https://www.atlassian.com/continuous-delivery/continuous-integration/trunk-based-development
* https://learn.microsoft.com/en-us/devops/develop/how-microsoft-develops-devops

*Trunk-based development* (*desarrollo basado en el tronco*) maximiza la velocidad de integración, pero exige *CI/CD* maduro: cada *commit* debe pasar *quality gates* automatizados antes de integrarse. Esta combinación —*commits* frecuentes a una rama protegida más validación automática exhaustiva— es la que mejor soporta el modelo de *shift left* descrito en capítulos previos. Sin ese nivel de automatización, *trunk-based development* reproduce los riesgos del flujo centralizado.

## El *workflow* como decisión de seguridad.

La elección del *workflow* de repositorio no es neutral desde la perspectiva de la cadena de suministro. Los ataques vistos en capítulos previos —modificación de *Actions* de terceros, envenenamiento de caché, exfiltración de secretos— se hacen más difíciles de introducir cuando el repositorio exige que todo cambio pase por un *pull request* con *quality gates* automáticos antes de llegar a la rama principal.

En este curso se trabaja con una estrategia de *workflow* simple y defendible:

- ramas cortas con propósito único;
- *pull request* como único camino hacia la rama principal;
- *checks* de *CI* obligatorios como condición de *merge*;
- rama principal protegida contra *push* directo;
- revisión obligatoria por al menos un par antes del *merge*.

Esa combinación no es solo una política de colaboración: es la primera capa de control de la cadena de suministro. El cuaderno siguiente introduce *GitHub* como la plataforma donde esas políticas se implementan como restricciones técnicas verificables.
