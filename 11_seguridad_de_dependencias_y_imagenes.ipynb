# Seguridad de dependencias e imágenes.

Este cuaderno orienta la seguridad a dos frentes de este curso: dependencias del proyecto y superficie de ataque de las imágenes publicadas por el *pipeline*.

## Objetivos.

- Reforzar buenas prácticas de *hardening* en *Actions*.
- Relacionar severidad con políticas de bloqueo.
- Introducir el control de *actions* y dependencias de terceros.

## Recomendaciones base.

Retomamos varias prácticas de endurecimiento:

- no guardar secretos en texto plano;
- vigilar cambios a *workflows* con [CODEOWNERS](https://docs.github.com/en/repositories/managing-your-repositorys-settings-and-features/customizing-your-repository/about-code-owners);
- desconfiar de entradas no fiables e inyecciones de script;
- revisar *actions* de terceros y preferir versiones fijadas;
- usar *Dependabot* para mantener referencias actualizadas.

Una política de severidad útil distingue entre hallazgos que bloquean automáticamente y hallazgos que requieren análisis compensatorio.

## *SCA*: análisis de composición de software.

El análisis de composición de software (*SCA*, por sus siglas en inglés) consiste en identificar todos los componentes de terceros que forman parte de una aplicación y verificar si alguno tiene vulnerabilidades conocidas (*CVEs*).

En una aplicación *Python* moderna, esos componentes son tres capas:

```text
Imagen de contenedor
  ├── Sistema operativo base (ej. python:3.12-slim = Debian)
  │     └── paquetes del *OS*: openssl, libexpat, zlib, ...
  └── Aplicación
        └── dependencias Python: requests, apiflask, gunicorn, ...
```

Las dependencias de la aplicación incluyen el servidor *WSGI* [*gunicorn*](https://gunicorn.org/), que ejecuta el servicio en producción.

Cualquiera de esas capas puede contener un *CVE* publicado. La imagen base que funcionaba hace tres meses puede tener hoy vulnerabilidades críticas en sus paquetes de sistema, aunque el código de la aplicación no haya cambiado.

*SCA* en el *pipeline* responde a dos preguntas concretas:
- qué versión de cada componente estamos usando;
- si alguna versión tiene un *CVE* conocido con severidad relevante.

## *Trivy*: escáner unificado de vulnerabilidades.

[*Trivy*](https://github.com/aquasecurity/trivy) es un escáner de seguridad de código abierto desarrollado por *Aqua Security*. En un solo binario cubre los dos frentes de este capítulo: dependencias del proyecto e imágenes de contenedor.

Lo que *Trivy* puede escanear:

| Objetivo | Qué detecta |
|---|---|
| Imagen de contenedor | *CVEs* en paquetes del *OS* (*apt*, *apk*) y en dependencias de lenguaje |
| Sistema de archivos / lockfile | *CVEs* en paquetes *Python*, *Node*, *Go*, etc. declarados en el proyecto |
| Repositorio *git* | Secretos expuestos, configuraciones inseguras, *IaC* mal configurada |
| Dockerfile | Malas prácticas de construcción (también cubre parte de lo que hace *Hadolint*) |

*Trivy* consulta múltiples bases de datos de vulnerabilidades: *NVD* (*NIST*), *GitHub Advisory Database*, y bases específicas por distribución (Alpine, Debian, RHEL). Las actualiza automáticamente antes de cada escaneo.

La diferencia con *Dependabot* es el alcance: *Dependabot* actualiza versiones; *Trivy* informa sobre vulnerabilidades conocidas en las versiones que ya tienes, incluyendo las del sistema operativo base de la imagen.

## *Trivy*: comandos principales.

*Trivy* opera en varios modos según el objetivo del escaneo:

```bash
# escanear una imagen de contenedor
trivy image mi-app:latest

# escanear solo vulnerabilidades CRITICAL y HIGH, salir con error si hay hallazgos
trivy image --severity CRITICAL,HIGH --exit-code 1 mi-app:latest

# escanear el sistema de archivos del proyecto (dependencias, configs, secretos)
trivy fs .

# escanear el lockfile directamente
trivy fs --scanners vuln uv.lock

# generar reporte en formato *SARIF* (compatible con *GitHub Security*)
trivy image --format sarif --output resultado.sarif mi-app:latest

# escanear un repositorio remoto
trivy repo https://github.com/org/repo
```

El flag `--exit-code 1` es el que convierte a *Trivy* en un *quality gate* real: el comando retorna código de salida 1 si hay hallazgos con la severidad especificada, lo que hace fallar el paso del *workflow* y detiene el *pipeline*.

## *Trivy* en *GitHub Actions*.

La integración recomendada usa la *action* oficial de *Aqua Security*. El *job* escanea la imagen recién construida y falla si encuentra vulnerabilidades de severidad `CRITICAL` o `HIGH`.

Las *actions* se referencian por SHA de commit en lugar de por tag (ver sección *Dependabot*):

```yaml
jobs:
  scan:
    runs-on: ubuntu-latest
    steps:
      - # actions/checkout@v6
        uses: actions/checkout@de0fac2e4500dabe0009e67214ff5f5447ce83dd

      - name: Construir imagen
        run: docker build -t mi-app:${{ github.sha }} .

      - name: Escanear imagen con Trivy
        # aquasecurity/trivy-action@v0.36.0
        uses: aquasecurity/trivy-action@ed142fd0673e97e23eac54620cfb913e5ce36c25
        with:
          image-ref: mi-app:${{ github.sha }}
          format: table
          exit-code: 1
          severity: CRITICAL,HIGH
          ignore-unfixed: true

      - name: Escanear dependencias del proyecto
        # aquasecurity/trivy-action@v0.36.0
        uses: aquasecurity/trivy-action@ed142fd0673e97e23eac54620cfb913e5ce36c25
        with:
          scan-type: fs
          scan-ref: .
          format: sarif
          output: trivy-results.sarif

      - name: Subir resultados a GitHub Security
        # github/codeql-action/upload-sarif@v4
        uses: github/codeql-action/upload-sarif@7211b7c8077ea37d8641b6271f6a365a22a5fbfa
        with:
          sarif_file: trivy-results.sarif
```

El formato `sarif` permite ver los hallazgos directamente en la pestaña **Security > Code scanning** del repositorio en *GitHub*, sin salir de la plataforma.

## *CodeQL* y el formato *SARIF*.

### Qué es *CodeQL*.

[*CodeQL*](https://codeql.github.com/) es el motor de análisis estático de seguridad de *GitHub*. A diferencia de *Trivy*, que consulta bases de datos de *CVEs* conocidos en dependencias, *CodeQL* analiza la **semántica del código fuente**: construye un modelo del flujo de datos del programa y ejecuta consultas sobre él para detectar patrones de vulnerabilidad como inyección *SQL*, *XSS*, deserialización insegura o uso de criptografía débil.

| | *Trivy* | *CodeQL* |
|---|---|---|
| Qué analiza | Dependencias y configuración | Código fuente |
| Cómo detecta | CVEs en bases de datos públicas | Flujo de datos y patrones semánticos |
| Falsos positivos | Bajos (versión conocida = CVE conocido) | Posibles (depende de la consulta) |
| Disponibilidad | Open source | Gratuito en repos públicos; *GHAS* en privados |

### El formato *SARIF*.

*SARIF* (*Static Analysis Results Interchange Format*) es el estándar abierto (*ISO/IEC 5055*) para intercambiar resultados de herramientas de análisis estático. Cualquier herramienta que produzca un archivo `.sarif` puede publicar sus hallazgos en la pestaña **Security > Code scanning** de *GitHub* usando la *action* `github/codeql-action/upload-sarif`.

Esto es exactamente lo que hace el ejemplo de *Trivy* de la sección anterior: *Trivy* no es *CodeQL*, pero produce *SARIF* y usa la infraestructura de *Code scanning* de *GitHub* para mostrar sus resultados. La *action* `upload-sarif` es el puente, no el analizador.

```text
Trivy fs . --format sarif  →  trivy-results.sarif
                                      │
                           upload-sarif action
                                      │
                           GitHub Security > Code scanning
```

### *CodeQL* como *quality gate*.

Cuando se configura *CodeQL* directamente (no solo el `upload-sarif`), *GitHub* ejecuta el análisis en cada *pull request* y bloquea el *merge* si detecta vulnerabilidades nuevas respecto a la rama base. La configuración mínima usa el workflow predeterminado:

```yaml
name: CodeQL

on:
  push:
    branches: ["main"]
  pull_request:
    branches: ["main"]

jobs:
  analyze:
    runs-on: ubuntu-latest
    permissions:
      contents: read
      security-events: write

    steps:
      - # actions/checkout@v6
        uses: actions/checkout@de0fac2e4500dabe0009e67214ff5f5447ce83dd

      - name: Inicializar CodeQL
        # github/codeql-action/init@v4
        uses: github/codeql-action/init@7211b7c8077ea37d8641b6271f6a365a22a5fbfa
        with:
          languages: python

      - name: Ejecutar análisis CodeQL
        # github/codeql-action/analyze@v4
        uses: github/codeql-action/analyze@7211b7c8077ea37d8641b6271f6a365a22a5fbfa
```

*GitHub* también ofrece *CodeQL* con configuración automática (*default setup*) activable directamente desde `Settings → Code security → Code scanning` sin necesidad de añadir el workflow manualmente.

## *Dependabot*: actualizaciones automáticas de dependencias.

[*Dependabot*](https://docs.github.com/en/code-security/dependabot) es una funcionalidad de *GitHub* que abre *pull requests* automáticamente cuando detecta versiones más recientes de las dependencias declaradas en el proyecto.

### Pinning por SHA: el control de cadena de suministro.

Las *actions* de terceros se referencian habitualmente por tag:

```yaml
uses: actions/checkout@v6
```

El problema es que **los tags son mutables**: el mantenedor puede moverlos silenciosamente a un commit diferente, y el *pipeline* ejecutaría código distinto al que se revisó originalmente sin ninguna advertencia. Este vector de ataque se conoce como *poisoned pipeline execution*.

La solución es referenciar la *action* por el SHA inmutable del commit, con el tag como comentario para mantener la legibilidad:

```yaml
# actions/checkout@v6
uses: actions/checkout@de0fac2e4500dabe0009e67214ff5f5447ce83dd
```

Un SHA sin comentario es imposible de mantener. Un tag sin SHA no ofrece garantía de inmutabilidad. Los dos juntos son el control completo.

### El par completo: SHA + *Dependabot*.

SHA pinning sin *Dependabot* convierte las actualizaciones en deuda manual: alguien tiene que notar que existe una nueva versión, buscar el SHA del nuevo commit y actualizar el archivo. *Dependabot* automatiza ese proceso: cuando el mantenedor publica una nueva versión, abre un *PR* actualizando el SHA **y** el comentario de tag a la vez:

```yaml
# Antes:
uses: actions/checkout@de0fac2e4500dabe0009e67214ff5f5447ce83dd  # v6

# PR de Dependabot lo actualiza a:
uses: actions/checkout@<sha-de-v7>  # v7
```

Ninguno de los dos controles funciona correctamente sin el otro: el SHA aísla el riesgo de tags mutables; *Dependabot* garantiza que ese SHA no quede obsoleto indefinidamente.

### Configuración de `.github/dependabot.yml`.

```yaml
version: 2
updates:
  - package-ecosystem: "pip"
    directory: "/"
    schedule:
      interval: "weekly"
    open-pull-requests-limit: 5
    groups:
      dev-dependencies:
        patterns: ["pytest*", "ruff", "mypy"]

  - package-ecosystem: "github-actions"
    directory: "/"
    schedule:
      interval: "weekly"
    open-pull-requests-limit: 5
    groups:
      # Agrupa actions relacionadas en un solo PR para reducir ruido de revisión.
      gcp-actions:
        patterns: ["google-github-actions/*"]
      security-actions:
        patterns: ["aquasecurity/*"]
```

El bloque `github-actions` cubre todos los workflows del repositorio. Los `groups` reducen el ruido agrupando *actions* relacionadas en un solo *PR*.

Cada *PR* de *Dependabot* pasa por el mismo *pipeline* de *CI* que cualquier otro cambio, garantizando que la actualización no rompa el proyecto antes de hacer *merge*.

## *uv* y la cadena de seguridad de dependencias.

`uv` contribuye a la seguridad de dependencias de dos formas complementarias:

**Lockfile con hashes verificados.** `uv.lock` incluye el hash `SHA-256` de cada paquete descargado. Al instalar con `uv sync --frozen`, *uv* verifica que cada archivo descargado coincida exactamente con el hash registrado. Esto hace imposible un ataque de sustitución silenciosa de paquetes.

**Reproducibilidad auditada.** Dado que `uv.lock` es un archivo de texto versionado en *git*, cualquier cambio en las dependencias produce un *diff* visible en el *pull request*. El equipo puede revisar qué se agregó o actualizó antes de aprobar el *merge*.

```text
pyproject.toml   <- intenciones: "requests >= 2.31"
uv.lock          <- realidad fijada: requests==2.32.3 sha256=abc123...
CI               <- uv sync --frozen verifica hashes antes de instalar
```

*Trivy* puede escanear directamente el `uv.lock` para detectar *CVEs* en las versiones fijadas, cerrando el ciclo entre gestión de dependencias y seguridad.

## Política de severidad.

No todos los hallazgos tienen el mismo peso. Una política útil clasifica los resultados por severidad y define qué nivel de hallazgo detiene automáticamente el *pipeline*:

| Severidad | Acción recomendada |
|---|---|
| `CRITICAL` | Bloqueo inmediato. El *pipeline* no avanza. |
| `HIGH` | Bloqueo inmediato. Requiere remediación antes del *merge*. |
| `MEDIUM` | *Warning*. Se registra pero no bloquea; debe resolverse en el sprint. |
| `LOW` | Informativo. Se reporta para seguimiento. |
