# *GitHub*.

Como se vio en capítulos previos, los *workflows* de repositorio son la primera capa de control de la cadena de suministro. *GitHub* es la plataforma donde esas políticas se implementan como restricciones técnicas verificables, y donde vive el pipeline de *CI/CD* que ejecuta los *quality gates*.

En el contexto de este curso, *GitHub* no es solo un servicio de hospedaje de repositorios: es la infraestructura que conecta el control de acceso al código, las políticas de revisión, la automatización del pipeline y una parte significativa de la superficie de ataque descrita en capítulos previos.

## Objetivos de la sesión

Al finalizar esta sesión podrás:

- describir los componentes clave de *GitHub* para colaboración, control de cambios y automatización;
- relacionar *pull requests*, ramas protegidas y reglas de repositorio con controles de seguridad del pipeline;
- distinguir qué capacidades de *GitHub* impactan directamente en la postura *DevSecOps* del equipo.

## Servicios de repositorios remotos de *Git*.

Debido a la popularidad de *Git*, muchos proveedores ofrecen servicios de hospedaje de repositorios remotos. Entre los más relevantes:

* [*GitHub*](https://github.com/) — El más popular, propiedad de Microsoft desde 2018.
* [*GitLab*](https://about.gitlab.com/) — Ofrece una versión *self-hosted* y una nube pública.
* [*Bitbucket*](https://bitbucket.org/) — Integrado con el ecosistema *Atlassian* (*Jira*, *Confluence*).
* [*Gitea*](https://gitea.io/) / [*Forgejo*](https://forgejo.org/) — Alternativas ligeras de código abierto.

Todos ofrecen hospedaje gratuito básico, pero se diferencian en sus funcionalidades avanzadas de gestión de proyectos, seguridad y *DevOps*.

## El servicio *GitHub*.

[*GitHub*](https://github.com/) es la plataforma de desarrollo colaborativo más utilizada del mundo, con más de 100 millones de desarrolladores registrados. Además del hospedaje de repositorios *Git*, ofrece un ecosistema completo de herramientas:

### Gestión de código
* **Repositorios** públicos y privados ilimitados.
* ***Pull Requests*** con revisión de código, comentarios en línea y aprobaciones.
* **Ramas protegidas** (*protected branches*) y reglas de fusión.
* ***Code owners*** para asignación automática de revisores.
* ***GitHub Gists*** — fragmentos de código compartibles.

### *CI/CD* y automatización
* **[*GitHub Actions*](https://github.com/features/actions)** — flujos de trabajo de *CI/CD* definidos en *YAML*, con miles de acciones reutilizables en el [*Marketplace*](https://github.com/marketplace).
* ***Runners*** propios (*Ubuntu, Windows, macOS*) o auto-hospedados (*self-hosted*).
* ***Environments*** con reglas de despliegue y secretos por entorno.

### Seguridad
* **[*Dependabot*](https://github.com/features/security)** — alertas y actualizaciones automáticas de dependencias vulnerables.
* ***Secret scanning*** — detección de credenciales expuestas en el código.
* **[*Code scanning / CodeQL*](https://codeql.github.com/)** — análisis estático para detectar vulnerabilidades.
* **[*GitHub Advanced Security*](https://docs.github.com/en/get-started/learning-about-github/about-github-advanced-security)** (licencia separada) — herramientas de seguridad avanzadas para repositorios privados.

### Inteligencia artificial
* **[*GitHub Copilot*](https://github.com/features/copilot)** — asistente de código basado en IA (modelos *GPT-4o, Claude Sonnet, Gemini*), disponible en el editor, en la línea de comandos y como agente autónomo (*Copilot Workspace*).

### Publicación y distribución
* **[*GitHub Pages*](https://pages.github.com/)** — hospedaje gratuito de sitios estáticos directamente desde un repositorio.
* **[*GitHub Packages*](https://github.com/features/packages)** — registro de paquetes compatible con *npm*, *Docker*, *Maven*, *NuGet*, *RubyGems* y *Gradle*.
* **[*GitHub Releases*](https://docs.github.com/en/repositories/releasing-projects-on-github)** — publicación de versiones con binarios adjuntos.

### Entornos de desarrollo en la nube
* **[*GitHub Codespaces*](https://github.com/features/codespaces)** — entornos de desarrollo completos basados en contenedores, accesibles desde el navegador o VS Code.

### Gestión de proyectos y comunidad
* **[*GitHub Projects*](https://github.com/features/issues)** — tableros kanban y vistas de tabla para planificación de proyectos.
* **[*Issues*](https://github.com/features/issues)** con etiquetas, milestones y asignados.
* **[*GitHub Discussions*](https://github.com/features/discussions)** — foros de comunidad por repositorio u organización.
* **[*GitHub Sponsors*](https://github.com/sponsors)** — financiamiento directo a mantenedores de proyectos open source.

## Las funcionalidades de seguridad de *GitHub* en contexto.

Las herramientas de seguridad de *GitHub* no son funcionalidades opcionales de conveniencia: responden directamente a los vectores de ataque vistos en capítulos previos.

- ***Dependabot*** responde al vector de dependencias con CVE: monitorea las dependencias del proyecto y abre pull requests automáticamente cuando detecta versiones con vulnerabilidades conocidas. Es la primera línea de defensa automatizada frente a incidentes del tipo *Log4Shell* o *XZ Utils*.
- ***Secret scanning*** responde al vector de secretos expuestos: detecta credenciales, tokens y llaves privadas que alguien haya commitido accidentalmente, incluyendo el historial completo del repositorio.
- ***Code scanning / CodeQL*** aplica análisis estático como *quality gate* de seguridad sobre el código fuente: parte del *shift left* descrito en capítulos previos antes de que el cambio llegue a producción.
- ***Environments con protección*** implementa el modelo de confianza verificada de capítulos previos: secretos acotados por entorno, aprobaciones manuales para despliegues sensibles, y restricción de qué ramas pueden activar un despliegue a producción.
- ***GitHub Actions*** es el motor del pipeline *CI/CD* de este curso. Como se vio en capítulos previos, es también un vector de ataque si las *Actions* de terceros no se fijan por *SHA* inmutable.

## Planes de *GitHub*.

*GitHub* ofrece planes *Free*, *Pro*, *Team* y *Enterprise*. Para los efectos de este curso, los límites relevantes son los minutos de Actions (2 000/mes en *Free*, 3 000 en *Pro*/*Team*) y la disponibilidad de funcionalidades de seguridad avanzadas: ramas protegidas con revisores requeridos y *Code owners* requieren *Team* o superior; *CodeQL* y *Advanced Security* están disponibles de forma gratuita en repositorios públicos.

Los precios y límites actualizados están en [github.com/pricing](https://github.com/pricing).

### *Forks*.

Un *fork* es una copia completa de un repositorio en la cuenta de otro usuario u organización. Es el mecanismo central del flujo de colaboración en *GitHub* con proyectos de código abierto:

1. Se hace *fork* del repositorio original.
2. Se trabaja en la copia propia con total libertad.
3. Se proponen los cambios al proyecto original mediante un *Pull Request*.

Los *forks* mantienen una referencia al repositorio original (*upstream*), lo que permite sincronizar cambios posteriores con:

```bash
git remote add upstream <URL-del-original>
git fetch upstream
git merge upstream/main
```

### *Pull Requests* (*PR*).

Un *Pull Request* es una solicitud formal para que el mantenedor de un repositorio integre cambios propuestos desde otra rama o *fork*. En el contexto *DevSecOps*, el *PR* no es solo una herramienta de colaboración: es el **punto de control central** de la cadena de entrega. Ningún cambio debería llegar a la rama principal sin pasar por este punto.

**Flujo típico:**

1. Crear una rama con los cambios (`git switch -c mi-feature`).
2. Hacer *push* de la rama al repositorio remoto.
3. Abrir un *PR* en *GitHub* indicando la rama de origen y destino.
4. Los *checks* de *CI/CD* se ejecutan automáticamente sobre el código del *PR*.
5. Los revisores comentan, solicitan cambios o aprueban.
6. Una vez aprobado y con todos los *checks* en verde, se fusiona en la rama destino.

**Características relevantes para la seguridad:**

* Revisiones requeridas (*required reviewers*) — ningún código llega a `main` sin aprobación humana.
* Integración con *GitHub Actions* — los *quality gates* se ejecutan como condición de *merge*.
* *Draft Pull Requests* — el cambio es visible al equipo desde el inicio, reduciendo el riesgo de trabajo en silos.
* Vinculación automática con *Issues* (`Closes #123` en la descripción cierra el *issue* al fusionar).

## Ramas protegidas como control de seguridad.

Las ramas protegidas (*protected branches*) son el mecanismo que convierte las políticas del *workflow* en restricciones técnicas verificables. Sin protección de ramas, las políticas de revisión son sugerencias que cualquier colaborador con acceso puede ignorar.

Una configuración mínima para un repositorio de este curso incluye:

- requerir *pull request* antes de hacer *merge* a `main`;
- requerir al menos una aprobación de revisión;
- requerir que los *checks* de *CI* pasen antes del *merge* (*required status checks*);
- bloquear el *push* directo a la rama principal.

El archivo `.github/CODEOWNERS` añade una capa adicional: asigna revisores automáticamente según el archivo modificado. Es especialmente relevante para los archivos de *workflow*, que son un vector de ataque crítico. Con esta configuración, cualquier cambio en los *workflows* requiere revisión explícita de un miembro del equipo de seguridad:

```
# Cualquier cambio en los workflows requiere revisión explícita
.github/workflows/  @equipo-seguridad
```

Esta configuración cierra un vector concreto de capítulos previos: un atacante que logre acceso al repositorio no puede modificar los archivos de *workflow* sin que un revisor autorizado lo apruebe explícitamente. La combinación de rama protegida + *checks* obligatorios + CODEOWNERS sobre *workflows* es la base técnica del *workflow* que se propone en este curso.
